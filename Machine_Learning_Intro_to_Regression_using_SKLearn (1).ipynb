```python
from sklearn.datasets import fetch_california_housing
import pandas as pd

california_data = fetch_california_housing()

df = pd.DataFrame(california_data.data, columns=california_data.feature_names)
df['target'] = california_data.target
df.head()
```

```python
df.info()
```

```python
df.describe().T
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
df.hist(figsize=(15,10), bins=50)
plt.show()
```

```python
X = df.drop('target', axis=1)
y = df['target']
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=77)
```

```python
X_train
```

```python
from sklearn.linear_model import LinearRegression

model_lr = LinearRegression()
model_lr.fit(X_train, y_train)
```

```python
model_lr.coef_
```

```python
model_lr.intercept_
```

```python
y_pred_lr = model_lr.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error

mse_lr = mean_squared_error(y_test, y_pred_lr)
mae_lr = mean_absolute_error(y_test, y_pred_lr)

print("Mean Squared Error (MSE):", mse_lr)
print("Mean Absolute Error (MAE):", mae_lr)
```

```python
from numpy import random
from sklearn.linear_model import SGDRegressor #stochastic gredient descent

model_sgd = SGDRegressor(
    max_iter=10000,
    eta0=0.85,
    learning_rate='adaptive', #constant, adaptive
    random_state=77,
    tol=1e-3
)

model_sgd.fit(X_train, y_train)
```

```python
y_pred_sgd = model_sgd.predict(X_test)

mse_sgd = mean_squared_error(y_test, y_pred_sgd)
mae_sgd = mean_absolute_error(y_test, y_pred_sgd)

print("Mean Squared Error (MSE):", mse_sgd)
print("Mean Absolute Error (MAE):", mae_sgd)
```

```python
X
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)

X_train_poly, X_test_poly, y_train, y_test = train_test_split(X_poly, y, test_size=0.2, random_state=77)

model_lr_poly = LinearRegression()
model_lr_poly.fit(X_train_poly, y_train)
```

```python

model_lr_poly.predict(X_train_poly)
mse_lr_poly = mean_squared_error(y_train, model_lr_poly.predict(X_train_poly))
mae_lr_poly = mean_absolute_error(y_train, model_lr_poly.predict(X_train_poly))

print("Mean Squared Error (MSE):", mse_lr_poly)
print("Mean Absolute Error (MAE):", mae_lr_poly)
```

```python
model_lr_poly.predict(X_test_poly)
mse_lr_poly = mean_squared_error(y_test, model_lr_poly.predict(X_test_poly))
mae_lr_poly = mean_absolute_error(y_test, model_lr_poly.predict(X_test_poly))

print("Mean Squared Error (MSE):", mse_lr_poly)
print("Mean Absolute Error (MAE):", mae_lr_poly)
```

```python
from sklearn.linear_model import Ridge

model_ridge = Ridge(alpha=0.01)
model_ridge.fit(X_train, y_train)
y_pred_ridge = model_ridge.predict(X_test)

mse_ridge = mean_squared_error(y_test, y_pred_ridge)
mae_ridge = mean_absolute_error(y_test, y_pred_ridge)

print("Mean Squared Error (MSE):", mse_ridge)
print("Mean Absolute Error (MAE):", mae_ridge)
```

```python
from sklearn.linear_model import Ridge

model_ridge = Ridge(alpha=0.7)
model_ridge.fit(X_train_poly, y_train)
y_pred_ridge = model_ridge.predict(X_test_poly)

mse_ridge = mean_squared_error(y_test, y_pred_ridge)
mae_ridge = mean_absolute_error(y_test, y_pred_ridge)

print("Mean Squared Error (MSE):", mse_ridge)
print("Mean Absolute Error (MAE):", mae_ridge)
```

```python
from sklearn.linear_model import Lasso

model_lasso = Lasso(alpha=0.1)
model_lasso.fit(X_train_poly, y_train)
y_pred_lasso = model_lasso.predict(X_test_poly)

mse_lasso = mean_squared_error(y_test, y_pred_lasso)
mae_lasso = mean_absolute_error(y_test, y_pred_lasso)

print("Mean Squared Error (MSE):", mse_lasso)
print("Mean Absolute Error (MAE):", mae_lasso)
```
